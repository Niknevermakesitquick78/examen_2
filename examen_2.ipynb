```python
import random

class Mazo:
    def __init__(self):
        self.cartas = [2,3,4,5,6,7,8,9,10,10,10,10,11]

    def dar_carta(self):
        return random.choice(self.cartas)


class Jugador:
    def __init__(self, nombre):
        self.nombre = nombre
        self.cartas = []
        self.puntos = 0

    def pedir_carta(self, mazo):
        carta = mazo.dar_carta()
        self.cartas.append(carta)
        self.calcular_puntos()
        print("Carta recibida beba:🃏", carta)

    def calcular_puntos(self):
        total = 0
        for carta in self.cartas:
            total += carta
        self.puntos = total

    def mostrar_mano(self):
        print("----------------------")
        print("Jugador:", self.nombre)
        print("Cartas:", self.cartas)
        print("Puntos 🍒:", self.puntos)
        print("----------------------")


mazo = Mazo()
jugador = Jugador("Jugador Guapo")
bartender = Jugador("Bartender🍸")

for _ in range(2):
    jugador.pedir_carta(mazo)
    bartender.pedir_carta(mazo)

jugador.mostrar_mano()
print("Carta visible del bartender:", bartender.cartas[0])

while True:
    if jugador.puntos > 21:
        print("Te pasaste de 21 mi so. Perdiste vuelve a intentarlo 🃏, gano el bartender.")
        break

    if jugador.puntos == 21:
        print("Blackjack. Ganaste mi so.")
        break

    opcion = input("¿Pedir (p) o plantarse (s)?: ").lower()

    while opcion not in ["p", "s"]:
        print("Entrada inválida")
        opcion = input("Ingresa 'p' o 's': ").lower()

    if opcion == "p":
        jugador.pedir_carta(mazo)
        jugador.mostrar_mano()
    else:
        print("Te plantaste con", jugador.puntos)
        break


if jugador.puntos <= 21:
    print("Turno del bartender")
    bartender.mostrar_mano()

    while bartender.puntos < 17:
        bartender.pedir_carta(mazo)
        bartender.mostrar_mano()

    if bartender.puntos > 21:
        print("El bartender se pasó, ganas mi so 🃏")
    else:
        print("Resultado final")
        jugador.mostrar_mano()
        bartender.mostrar_mano()

        if jugador.puntos > bartender.puntos:
            print("Ganaste mi so 🃏")
        elif jugador.puntos < bartender.puntos:
            print("El bartender gana 🍒")
        else:
            print("Empate 🍒")
```
