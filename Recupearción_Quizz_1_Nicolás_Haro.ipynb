```python
#Recuperación del quizz 1 Nicolás Haro
class Tablero:
    def __init__(self):
        self.casillas = {1:" ", 2:" ", 3:" ",
                         4:" ", 5:" ", 6:" ",
                         7:" ", 8:" ", 9:" "}

    def mostrar(self):
        print()
        print(f" {self.casillas[1]} | {self.casillas[2]} | {self.casillas[3]} ")
        print("---+---+---")
        print(f" {self.casillas[4]} | {self.casillas[5]} | {self.casillas[6]} ")
        print("---+---+---")
        print(f" {self.casillas[7]} | {self.casillas[8]} | {self.casillas[9]} ")
        print()

    def marcar(self, posicion, simbolo):
        if self.casillas[posicion] == " ":
            self.casillas[posicion] = simbolo
            return True
        return False

    def hay_ganador(self, simbolo):
        combinaciones = [
            (1,2,3),(4,5,6),(7,8,9),
            (1,4,7),(2,5,8),(3,6,9),
            (1,5,9),(3,5,7)
        ]
        for a,b,c in combinaciones:
            if self.casillas[a] == simbolo and self.casillas[b] == simbolo and self.casillas[c] == simbolo:
                return True
        return False

    def lleno(self):
        for valor in self.casillas.values():
            if valor == " ":
                return False
        return True


class Jugador:
    def __init__(self, nombre, simbolo):
        self.nombre = nombre
        self.simbolo = simbolo

    def jugada(self, tablero):
        while True:
            try:
                posicion = int(input(f"{self.nombre} ({self.simbolo}) elige casilla (1-9): "))
                if posicion in tablero.casillas:
                    if tablero.marcar(posicion, self.simbolo):
                        break
                    else:
                        print("Casilla ocupada")
                else:
                    print("Posición inválida")
            except:
                print("Entrada inválida")


tablero = Tablero()
jugador1 = Jugador("Jugador 1", "X")
jugador2 = Jugador("Jugador 2", "O")

turno = 0

while True:
    tablero.mostrar()

    if turno % 2 == 0:
        jugador1.jugada(tablero)
        if tablero.hay_ganador(jugador1.simbolo):
            tablero.mostrar()
            print("Gana", jugador1.nombre)
            break
    else:
        jugador2.jugada(tablero)
        if tablero.hay_ganador(jugador2.simbolo):
            tablero.mostrar()
            print("Gana", jugador2.nombre)
            break

    if tablero.lleno():
        tablero.mostrar()
        print("Empate")
        break

    turno += 1
```

```python
#Tercer ejercicio:
import random

class Maquina:
    def __init__(self):
        self.simbolos = ["🍒", "7", "💎", "⭐"]
        self.factor_dificultad = 1

    def tirar(self):
        resultado = []
        for _ in range(3):
            prob_ganar = 0.6 / self.factor_dificultad
            if random.random() < prob_ganar:
                simbolo = random.choice(self.simbolos)
            else:
                simbolo = random.choice(self.simbolos)
            resultado.append(simbolo)
        self.factor_dificultad += 0.2
        return resultado

    def evaluar(self, resultado):
        if resultado[0] == resultado[1] == resultado[2]:
            return 50
        elif resultado[0] == resultado[1] or resultado[1] == resultado[2] or resultado[0] == resultado[2]:
            return 20
        else:
            return 0


class Jugador:
    def __init__(self, nombre, monedas):
        self.nombre = nombre
        self.monedas = monedas

    def apostar(self):
        if self.monedas >= 10:
            self.monedas -= 10
            return True
        return False

    def recibir_premio(self, cantidad):
        self.monedas += cantidad

    def mostrar_saldo(self):
        print("Monedas:", self.monedas)


maquina = Maquina()
jugador = Jugador("Jugador", 100)

while True:
    jugador.mostrar_saldo()

    if jugador.monedas < 10:
        print("No tienes suficientes monedas. Fin del juego.")
        break

    opcion = input("¿Tirar? (s/n): ").lower()

    while opcion not in ["s", "n"]:
        print("Entrada inválida")
        opcion = input("¿Tirar? (s/n): ").lower()

    if opcion == "n":
        print("Te retiraste del juego.")
        break

    if jugador.apostar():
        resultado = maquina.tirar()
        print("[", resultado[0], "] [", resultado[1], "] [", resultado[2], "]")

        premio = maquina.evaluar(resultado)

        if premio == 50:
            print("3 iguales. Ganas 50 monedas")
        elif premio == 20:
            print("2 iguales. Ganas 20 monedas")
        else:
            print("No ganaste")

        jugador.recibir_premio(premio)
```
